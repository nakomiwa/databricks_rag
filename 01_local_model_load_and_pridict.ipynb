```python
import mlflow

model_name = "bge_m3"
model_stage = "Production"  # またはバージョン番号 e.g. "1"


catalog = "system"        # 実際のカタログ名
schema  = "ai"          # 実際のスキーマ名
model_name = "bge_m3"
version = "1"           # またはエイリアス: "@champion"

# バージョン番号で指定
model_uri = f"models:/{catalog}.{schema}.{model_name}/{version}"

embedding_model = mlflow.pyfunc.load_model(model_uri)
# embedding_modelはテキストを入力してベクトルを出力するモデル
```

```python
import pandas as pd

# 1件だけ埋め込みたい場合
df = pd.DataFrame({"input": ["テスト"]})
embeddings = embedding_model.predict(df)

print(embeddings)
```
